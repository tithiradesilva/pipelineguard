```python
df = spark.read.option("header", "true").option("inferSchema", "true").csv(
    "abfss://raw@stpipelineguard01.dfs.core.windows.net/sales.csv"
)

display(df)
```

```python
df_with_duplicate = df.union(df.limit(1))

print("Rows before removing duplicates:", df_with_duplicate.count())
```

```python
df_clean = df_with_duplicate.dropDuplicates(["order_id"])

print("Rows after removing duplicates:", df_clean.count())
```

```python
from pyspark.sql.functions import col

df_transformed = df_clean.withColumn(
    "total_amount",
    col("quantity") * col("price")
)

display(df_transformed)
```

```python
output_path = "abfss://processed@stpipelineguard01.dfs.core.windows.net/sales_processed"

df_transformed.write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(output_path)

print("Processed data written successfully")
```
